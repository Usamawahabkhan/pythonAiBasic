```python
import matplotlib.pyplot as plt

```

```python
x = [1, 2, 3, 4, 5]
y = [10, 12, 5, 8, 15]
```

```python
plt.plot(x, y, marker='o', linestyle='--', color='b', label='Data Points')
plt.legend()
#plt.plot(x,y)
plt.title("Sample Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

```python
plt.plot(x, y, color='red', linestyle='--', marker='o')
plt.title("Customized Plot", fontsize=14)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend(['Data Line'])
plt.grid(True)
```

```python
import pandas as pd

# Load the car sales data
car_sales = pd.read_csv('./car-sales.csv')

```

```python
# Show a simple plot of the data (e.g., number of cars sold per brand)
car_sales['Make'].value_counts().plot(kind='bar')
plt.title("Number of Cars Sold by Brand")
plt.xlabel("Car Brand")
plt.ylabel("Number of Cars Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load built-in dataset
df = sns.load_dataset('titanic')

# Show first few rows
print(df.head())
```

```python
sns.get_dataset_names()
```

```python
sns.countplot(x='sex', data=df)

plt.title("count of Male vs Female")
plt.show()
```

```python
sns.histplot(df['age'], bins=30, kde=True)
plt.title("Age Distribution with KDE")
plt.show()
```

```python
sns.boxplot(x='sex', y='age', data=df)
plt.title("Age Distribution by Gender")
plt.show()
```

```python
sns.violinplot(x='class', y='age', data=df)
plt.title("Age Distribution by Class")
plt.show()
```

```python
# Select numeric columns only
numeric_df = df.select_dtypes(include='number')
correlation = numeric_df.corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df[['age', 'fare', 'pclass', 'survived']], hue='survived')
plt.suptitle("Pairplot of Numeric Features by Survival", y=1.02)
plt.show()
```
