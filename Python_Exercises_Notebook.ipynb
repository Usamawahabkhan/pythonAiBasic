## 🧭 Control Flow
Control flow allows you to dictate the order in which instructions are executed. This includes conditional statements and loops.

```python
# Exercise 1: Even or Odd
num = int(input("Enter a number: "))
if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# Exercise 2: FizzBuzz
for i in range(1, 21):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# Exercise 3: Find Prime Numbers
for num in range(2, 20):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```

## 📋 Lists
Lists are ordered, mutable collections of items.

```python
# Exercise 4: Reverse a List
my_list = [1, 2, 3, 4, 5]
reversed_list = my_list[::-1]
print(reversed_list)
```

```python
# Exercise 5: Remove Duplicates
nums = [1, 2, 2, 3, 4, 4, 5]
unique = list(set(nums))
print(unique)
```

```python
# Exercise 6: Sum of Elements
numbers = [10, 20, 30, 40]
total = sum(numbers)
print("Sum:", total)
```

## 📦 Tuples
Tuples are ordered, immutable collections of items.

```python
# Exercise 7: Tuple Unpacking
person = ("Alice", 25, "Engineer")
name, age, job = person
print(f"{name} is a {age}-year-old {job}.")
```

```python
# Exercise 8: Swap Values
a = 5
b = 10
a, b = b, a
print(a, b)
```

## 📚 Dictionaries
Dictionaries store key-value pairs and are mutable.

```python
# Exercise 9: Count Word Frequency
sentence = "hello world hello"
words = sentence.split()
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1
print(freq)
```

```python
# Exercise 10: Access and Update Dictionary
student = {"name": "John", "age": 20}
student["age"] = 21
student["grade"] = "A"
print(student)
```

## 🔢 Sets
Sets are unordered collections of unique items.

```python
# Exercise 11: Set Operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
```

```python
# Exercise 12: Remove Duplicates Using Set
items = ["apple", "banana", "apple", "orange"]
unique_items = set(items)
print(unique_items)
```
